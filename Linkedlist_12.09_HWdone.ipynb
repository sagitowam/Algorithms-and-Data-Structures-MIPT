**Связный список** – это одна из базовых структур данных. Каждый элемент (или узел) ее состоит из, собственно, **хранимых данных** и **ссылок** на соседние элементы. Односвязный список хранит ссылку только на следующий элемент в структуре, а двусвязный – на следующий и предыдущий. Такая организация данных позволяет им располагаться в **любой области памяти** (в отличие от, например, массива, все элементы которого должны располагаться в памяти один за другим).

```python
class Node: # Объявляем класс
    # Описать ф-ии которые можно выполнять с элементами класса (методы)
    def __init__(self, data, next = None):
        # Значение, которое тут хранится. У объекта класса Node есть аттрибут data (его собственная) + мы передаем еще этот элемент ему
        self.data = data
        # Адрес следующего узла
        self.next = next

a = Node(10)
```

```python
# Конструктор это метод который вызывается первым у данного класса при создании нового объекта
class LinkedList:
    def __init__(self):
        self.head = None

# Добавление элемента к списку
    def append(self, data): # В данном методе не надо прописывать head, так как у нас внутри этого класса есть доступ ко всем методам класса
        if self.head:
            curnode = self.head
            while curnode.next:
                curnode = curnode.next
            curnode.next = Node(data)
        else:
            self.head = Node(data) 

# Вывод списка
    def print(self): # Так не надо называть функцию, лучше не называть как глобальную ф-ию, а что-то типа print_list
        if self.head:
            curnode = self.head
            while curnode: # Если тут будет curnode.next то последний узел не выведется 
                print(curnode.data) # Печатаем каждый узел
                curnode = curnode.next
        else:
            print('empty list')

# Возвращение и удаление значений
    def delete(self, n):
        if self.head: # Сначала проверяем есть ли хоть что-то
            curnode = self.head
            count = 0
            prev = None
            while curnode:
                if count == n:
                    if prev:
                        prev.next = curnode.next
                    else:
                        self.head = curnode.next # Предыдущего узла может не быть только у головы
                    return curnode.data
                count += 1
                prev = curnode
                curnode = curnode.next

    def delete_v2(self, n):
        if self.head: # Сначала проверяем есть ли хоть что-то
            if n == 0:
                for_del = self.head
                self.head = self.head.next
                return for_del
            curnode = self.head
            count = 0
            prev = None
            while curnode:
                if count == n-1:
                    for_del = curnode.next
                    curnode.next = curnode.next.next
                    return for_del.data
                count += 1
                curnode = curnode.next

# Удаление на рекурсии
    def delete_v3(self, n):
        if self.head:
            if n == 0:
                for_del = self.head
                self.head = self.head.next
                return for_del.data
            return self._delete(self.head, 0, n)

    def _delete(self, node, count, n):
        if count == n-1 and node.next:
            for_del = node.next
            node.next = node.next.next
            return for_del.data
        count += 1
        return self._delete(node.next, count, n)

# Инсерция после какого-то элемента
    def insert(self, n, data):
        if self.head:
            if n == 0:
                curnode = self.head
                self.head = Node(data, curnode)
                return
            return self._insert(self.head, n, 0, data)
        
    def _insert(self, node, n, count, data):
        if node:
            if count == n - 1:
                curnode = node.next
                node.next = Node(data, curnode)
                return
            self._insert(node.next, n, count + 1, data)

    # Домашнее задание: разворот списка

    ## Итеративный разворот O(n)?
    def reverse(self):
        if self.head:
            curnode = self.head
            prevnode = None
            while curnode:
                nextnode = curnode.next
                curnode.next = prevnode
                prevnode = curnode
                curnode = nextnode
            self.head = prevnode
            return self.head

    ## Разворот с рекурсией O(n)?
    def reverse_rec(self, node = None):
        first = node is None
        if self.head:
            if node is None:
                node = self.head
            if node.next is None:
                return node
            newhead = self.reverse_rec(node.next) #Дальше вглубь
            node.next.next = node # тут смена связи
            node.next = None # тут разрыв связи
            if first:
                self.head = newhead
            return newhead


            
```
