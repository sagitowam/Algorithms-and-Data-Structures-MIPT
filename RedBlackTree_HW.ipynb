```python
class Node:
    def __init__(self, value, left = None, right = None, color = 'red'):
        self.value = value
        self.right = right
        self.left = left
        self.color = color
    
```

    Красно-чёрным называется бинарное поисковое дерево, у которого каждому узлу сопоставлен дополнительный атрибут — цвет и для которого выполняются следующие свойства:
    - Каждый узел промаркирован красным или чёрным цветом
    - Корень и конечные узлы (листья) дерева — чёрные: свойство с конечными узлами всегда верно, в ослабленном дереве корень может быть красным
    - У красного узла родительский узел — чёрный
    - Все простые пути из любого узла x до листьев содержат одинаковое количество чёрных узлов
    - Чёрный узел может иметь чёрного родителя

```python
class RBTree:
    def __init__(self):
        self.root = None # Только создали дерево, информации в нем нет

    def append(self, value):
        if self.root:
            self.root = self._append(self.root, value) # вход в рекурсию для вставки узла
        else:
            self.root = Node(value) 
        self.root.color = 'black' # делаю корень черным здесь, чтобы после балансировки корень оставался черным

    def _append(self, node, value):
        if node is None: # проверяем сам узел (чтобы вставлять влево и вправо без ошибки)
            return Node(value, color='red')
        if value > node.value: # проверяем в какую сторону вставлять узел
            node.right = self._append(node.right, value)
        elif value < node.value:
            node.left = self._append(node.left, value)
        else:
            return node
        node = self._balance(node) # вызываю балансировку по самому узлу, к которому присоединяли

        return node

    def find(self, value):
        curnode = self.root
        while curnode is not None:
            if curnode.value == value:
                return curnode
            if curnode.value < value:
                curnode = curnode.right
            else:
                curnode = curnode.left
        return None

    #def height(self):



    def rotation(self, node, direction):
        if direction == 'left':
            curnode = node.right # сохраняю правую часть в отдельный узел
            node.right = curnode.left # записываю левый узел от правой части к узлу, от которого поворачиваю
            curnode.left = node # делаю узел, относительно которого поворачиваю, подузлом узла, встающего на его место
            return curnode
        else: # тут все то же самое, только направо
            curnode = node.left
            node.left = curnode.right
            curnode.right = node
            return curnode   

    def is_red(self, node):
        return node is not None and node.color == 'red'

    #def is_black(self, node):
        #return node is not None and node.color == 'black'
    
    #def balance(self):
        #if self.root:
            #self.root = self._balance(self.root)

    
    def _balance(self, node):
        if self.is_red(node.left):
            if self.is_red(node.left.left):  # LL
                if self.is_red(node.right):
                    node.right.color = 'black'
                    node.left.color = 'black'
                    node.color = 'red' # тут окрашиваю родителя в красный (если узел является корнем, то мы все равно при аппенде вернем его черным)
                    return node
                else:
                    old_color = node.color # тут я сохраняю старый цвет, чтобы не менять исходную цветовую конструкцию при повороте
                    node = self.rotation(node, 'right') # поворот направо, потому что лево-лево
                    node.color = old_color # возвращаю все цвета на место
                    node.right.color = 'red'
                    return node

            elif self.is_red(node.left.right):  # LR
                if self.is_red(node.right):
                    node.right.color = 'black'
                    node.left.color = 'black'
                    node.color = 'red'
                    return node
                else:
                    old_color = node.color
                    node.left = self.rotation(node.left, 'left') # тут сначала поворот влево, чтобы получить позицию лево-лево
                    node = self.rotation(node, 'right') # а потом правый поворот
                    node.color = old_color
                    node.right.color = 'red'
                    return node

        if self.is_red(node.right):
            if self.is_red(node.right.right):  # RR - все то же самое но в другую сторону
                if self.is_red(node.left):
                    node.right.color = 'black'
                    node.left.color = 'black'
                    node.color = 'red'
                    return node
                else:
                    old_color = node.color
                    node = self.rotation(node, 'left')
                    node.color = old_color
                    node.left.color = 'red'
                    return node

            elif self.is_red(node.right.left):  # RL
                if self.is_red(node.left):
                    node.right.color = 'black'
                    node.left.color = 'black'
                    node.color = 'red'
                    return node
                else:
                    old_color = node.color
                    node.right = self.rotation(node.right, 'right')
                    node = self.rotation(node, 'left')
                    node.color = old_color
                    node.left.color = 'red'
                    return node

        return node


    def print_tree(self):
            self._print_tree_rec2(self.root, 0)
        
    def _print_tree_rec2(self, node, depth):
        if node:
            print('    ' * depth, node.value, node.color)
            self._print_tree_rec2(node.left, depth + 1)
            self._print_tree_rec2(node.right, depth + 1)
        
```
