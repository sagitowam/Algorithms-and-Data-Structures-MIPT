Алгоритм Белмана-Форда
Первая реализация O(V^3)

```python
def ford(g, start):
    dists = {}
    for i in range(len(g)):
        dists[i] = float('inf') # Определяю словарь расстояний
    dists[start] = 0 # от старта к старту = 0
    for i in range(len(g) - 1): # Прохожу количество вершин - 1
        for u in range(len(g)): # Цикл для первой вершины
            for v in range(len(g[u])): # Цикл для второй вершины
                if g[u][v] is not None: # Проверка на наличие ребра
                    if dists[v] > dists[u] + g[u][v]: # Проверка можем ли добраться до v более кратчайшим путем, если пойдем через u
                        dists[v] = dists[u] + g[u][v]
    for u in range(len(g)): # Проверка на отрицательный цикл
        for v in range(len(g[u])): 
            if g[u][v] is not None:
                if dists[v] > dists[u] + g[u][v]:
                    print("Negative weight cycle")
    return(dists)
```

Беллмана-Форда, но со списком ребер (O(VE)): алгоритм действий кода тот же самый

```python
class Graph:
    def __init__(self, v):
        self.V = v
        self.graph = []
    def edge(self, u, v, dist):
        self.graph.append([u, v, dist])
    def bell_ford(self, start):
        dists = [float('inf')] * self.V
        dists[start] = 0
        for i in range(self.V - 1):
            for u, v, dist in self.graph:
                if dists[u] != float('inf'):
                    if dists[v] > dists[u] + dist:
                        dists[v] = dists[u] + dist
        for u, v, dist in self.graph:
            if dists[u] != float('inf'):
                if dists[v] > dists[u] + dist:
                    print("Negative weight cycle")
                    return None
        return(dists)
```

```python
graph2 = Graph(5)
graph2.edge(0, 1, 1)
graph2.edge(1, 2, 2)
graph2.edge(1, 3, 3)
graph2.edge(0, 4, 4)
graph2.edge(0, 3, 5)
print(graph2.bell_ford(0))
```
